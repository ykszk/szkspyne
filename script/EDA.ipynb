```python
from pathlib import Path
import matplotlib.pyplot as plt
import json
from szkspyne import labelme_utils
import seaborn as sns
sns.set(context='notebook', style='whitegrid', font='serif')                    

json_dir = Path('/scratch/scoliosis/Osaka AI Team Dropbox/project/suzuki/scoliosis/annotation/completed')
```

```python
all_data = []
for fn in json_dir.glob('lateral_whole/*.json'):
    with open(fn) as f:
        data = json.load(f)
    data['filename'] = fn.relative_to(json_dir)
    all_data.append(data)
print(len(all_data))
```

```python
all_shapes = {}
for data in all_data:
    shapes = labelme_utils.conv2shape_dict(data)
    all_shapes[data['filename']] = shapes
shapes
```

```python
import pandas as pd
LABELS = ['TL','TR','BL','BR']
rows = []
for fn, shapes in all_shapes.items():
    points = shapes['point']
    row = [fn]
    for lbl in LABELS:
        row.append(len(points[lbl]))    
    rows.append(row)
df_counts = pd.DataFrame(rows, columns=['filename'] + LABELS)
df_counts
```

```python
df_less = df_counts[df_counts['TL']<19]
print(df_less.shape)
df_less
```
