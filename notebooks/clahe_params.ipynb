```python
import sys
sys.path.append('../szkspyne/')
import json
from pathlib import Path
data_dir = Path('../../data/annotation/annotated/completed')
with open(data_dir / 'valid_ap_whole.ndjson') as f:
    jsons = [json.loads(line) for line in f]
len(jsons)
```

```python
pops = [j for j in jsons if j['flags']['術後']]
pres = [j for j in jsons if not j['flags']['術後']]
len(pops), len(pres)
```

```python
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display
import transforms
import numpy as np

image_dir = Path('../../data/annotation/images/ap_whole')
pids = set()
# for data in pres[::-1]:
for data in pops[::-1]:
    if len(pids) >= 10:
        break
    image_path = image_dir / Path(data['imagePath'].replace('\\', '//')).name
    pid = image_path.name.split('_')[0]
    if pid in pids:
        continue
    else:
        pids.add(pid)
    img = Image.open(image_path)
    img.thumbnail((20000, 768))
    clahed = transforms.apply_clahe(np.array(img))
    clahed2 = transforms.apply_clahe(np.array(img), tileGridSize=(64, 64))
    sidebyside = Image.fromarray(np.concatenate((np.array(img), clahed, clahed2), axis=1))

    display(sidebyside)
```
